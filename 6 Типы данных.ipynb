## Part 6.2 SET & DICT

https://stepik.org/lesson/631917/step/17?unit=627943

```python
alphabet_translation = input().strip()
text_to_translate = input().strip()


translation_table = str.maketrans(
    'abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ',
    alphabet_translation.lower() + alphabet_translation.upper()
)

translated_text = text_to_translate.translate(translation_table)
print(translated_text.upper())
```
