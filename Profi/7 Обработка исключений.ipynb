```python
def swapcase_vowels(string):
    vowels = 'aeiouy'
    swapped_string = ''
    for char in string:
        if char in vowels:
            char = char.upper()
        swapped_string += char
    return swapped_string
```

```python
def get_max_index(numbers):
    max_index = 0
    max_value = 1

    for index, value in enumerate(numbers):
        if value > max_value:
            max_index = index
            max_value = value

    return max_index
```

```python
import calendar, locale
locale.setlocale(locale.LC_ALL, 'ru_RU.UTF-8')


def get_weekday(number):
    if not isinstance(number, int):
        raise TypeError('Аргумент не является целым числом')
    if number < 1 or number > 7:
        raise ValueError('Аргумент не принадлежит требуемому диапазону')
    else:
        return calendar.day_name[number-1]
```

```python
def get_id(names, name):
    if not isinstance(name, str):
        raise TypeError('Имя не является строкой')
    elif name.istitle() and name.isalpha():
        names.append(name)
        return len(names)
    raise ValueError ('Имя не является корректным')
```

```python
def is_good_password(string: str) -> bool:
    if len(string) < 9:
        return False
    has_lower = any(c.islower() for c in string)
    has_upper = any(c.isupper() for c in string)
    has_digit = any(c.isdigit() for c in string)
    return has_lower and has_upper and has_digit
```

```python
class PasswordError(Exception):
    pass

class LengthError(PasswordError):
    pass

class LetterError(PasswordError):
    pass

class DigitError(PasswordError):
    pass

def is_good_password(string):
    if len(string) < 9:
        raise LengthError
    if not any(c.isalpha() for c in string) or (string.islower() or string.isupper()):
        raise LetterError
    if not any(c.isdigit() for c in string):
        raise DigitError
    return True
```

```python
import sys
class PasswordError(Exception):
    pass

class LengthError(PasswordError):
    pass

class LetterError(PasswordError):
    pass

class DigitError(PasswordError):
    pass

def is_good_password(string):
    if len(string) < 9:
        raise LengthError('LengthError')
    if not all((any([i.isupper() for i in string]), any([i.islower() for i in string]))):
        raise LetterError('LetterError')
    if not any([i in string for i in '0123456789']):
        raise DigitError('DigitError')
    return True

for line in sys.stdin:
    try:
        if is_good_password(line.strip('\n')):
            print('Success!')
            break
    except Exception as e:
        print(e)
```
